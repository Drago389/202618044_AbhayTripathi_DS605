**Task 2 - Vectorized Arithmetic and Linear Algebra**

```python
import numpy as np

# ==========================================
# 1. Matrix Creation & Vectorized Arithmetic
# ==========================================
print("--- Matrix Arithmetic ---")
# Create two matrices of the same shape (2x2)
matrix_X = np.array([[1, 2],
                     [3, 4]])

matrix_Y = np.array([[5, 6],
                     [7, 8]])

# Addition
addition_result = matrix_X + matrix_Y
print(f"Addition (X + Y):\n{addition_result}\n")

# Element-wise multiplication
element_wise_mult = matrix_X * matrix_Y
print(f"Element-wise Multiplication (X * Y):\n{element_wise_mult}\n")

# Matrix multiplication (Dot product)
matrix_mult = matrix_X @ matrix_Y  # equivalent to np.matmul(matrix_X, matrix_Y)
print(f"Matrix Multiplication (X @ Y):\n{matrix_mult}\n")


# ==========================================
# 2. Linear Algebra (Transpose, Det, Inverse)
# ==========================================
print("--- Linear Algebra Operations ---")
# Create a square matrix A
matrix_A = np.array([[4, 7],
                     [2, 6]])

# Find its transpose
transpose_A = matrix_A.T
print(f"Matrix A:\n{matrix_A}\n")
print(f"Transpose of A:\n{transpose_A}\n")

# Find the determinant
det_A = np.linalg.det(matrix_A)
print(f"Determinant of A: {det_A:.2f}\n")

# Check if invertible (determinant is not zero), find inverse, and verify
if det_A != 0:
    # Find the inverse
    inverse_A = np.linalg.inv(matrix_A)
    print(f"Inverse of A (A^-1):\n{inverse_A}\n")

    # Verify A @ A^-1 equals the Identity Matrix
    identity_matrix = np.eye(matrix_A.shape[0])
    verification = matrix_A @ inverse_A

    # Use np.allclose to handle floating-point inaccuracies
    is_valid = np.allclose(verification, identity_matrix)

    print(f"A @ A^-1:\n{verification}")
    print(f"Verification successful (A @ A^-1 == I): {is_valid}")
else:
    print("Matrix A is singular (determinant is 0) and cannot be inverted.")
```

**Task 5 - Filtering and Querying**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('train.csv')

# ==========================================
# Question 1: Male passengers older than 50
# ==========================================
# Using boolean indexing
q1 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print(f"1. Male passengers older than 50: {len(q1)}")

# ==========================================
# Question 2: Female first-class passengers
# ==========================================
# Using query()
q2 = df.query("Sex == 'female' and Pclass == 1")
q2_count = len(q2)
q2_survived_pct = q2['Survived'].mean() * 100

print(f"2. Female first-class passengers: {q2_count}")
print(f"   Percentage that survived: {q2_survived_pct:.2f}%")

# ==========================================
# Question 3: Age 20-40, Fare above overall median, and survived
# ==========================================
overall_median_fare = df['Fare'].median()

# Using boolean indexing
q3 = df[(df['Age'] >= 20) &
        (df['Age'] <= 40) &
        (df['Fare'] > overall_median_fare) &
        (df['Survived'] == 1)]
print(f"3. Age 20-40, Fare > overall median, and survived: {len(q3)}")

# ==========================================
# Question 4: Travelling alone, age < 30, and did not survive
# ==========================================
# Using query()
q4 = df.query("SibSp == 0 and Parch == 0 and Age < 30 and Survived == 0")
print(f"4. Travelling alone, age < 30, and did not survive: {len(q4)}")

# ==========================================
# Question 5: Embarked='S', Pclass 2 or 3, Fare above Southampton median
# ==========================================
# First, calculate the median fare specifically for Southampton ('S')
s_median_fare = df[df['Embarked'] == 'S']['Fare'].median()

# Using boolean indexing and isin() for Pclass
q5 = df[(df['Embarked'] == 'S') &
        (df['Pclass'].isin([2, 3])) &
        (df['Fare'] > s_median_fare)]
print(f"5. Embarked='S', Pclass 2 or 3, Fare > Southampton median: {len(q5)}")
```

**Task 6 - groupby() and Aggregation**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('train.csv')

print("--- Task 6: groupby() and Aggregation ---\n")

# ==========================================
# 1. Survival rate by Sex
# ==========================================
# mean() of the 'Survived' column (which is 0s and 1s) gives the survival rate
surv_by_sex = df.groupby('Sex')['Survived'].mean()
print("1. Survival Rate by Sex:")
print(surv_by_sex, "\n")

# ==========================================
# 2. Survival rate by Pclass
# ==========================================
surv_by_pclass = df.groupby('Pclass')['Survived'].mean()
print("2. Survival Rate by Pclass:")
print(surv_by_pclass, "\n")

# ==========================================
# 3. Average Age and Fare by Pclass
# ==========================================
age_fare_by_pclass = df.groupby('Pclass')[['Age', 'Fare']].mean()
print("3. Average Age and Fare by Pclass:")
print(age_fare_by_pclass, "\n")

# ==========================================
# 4. Passenger count and survival rate by Sex-Pclass
# ==========================================
# Grouping by multiple columns and using agg() for different metrics
sex_pclass_agg = df.groupby(['Sex', 'Pclass'])['Survived'].agg(
    Passenger_Count='count',
    Survival_Rate='mean'
)
print("4. Passenger Count and Survival Rate by Sex and Pclass:")
print(sex_pclass_agg, "\n")

# ==========================================
# 5. Passenger count, average Fare, and survival rate by Embarked
# ==========================================
# Applying different aggregations to different columns
embarked_agg = df.groupby('Embarked').agg(
    Passenger_Count=('PassengerId', 'count'),
    Average_Fare=('Fare', 'mean'),
    Survival_Rate=('Survived', 'mean')
)
print("5. Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_agg)
```

**Task 7 - Missing Values and Fare Outliers**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('train.csv')

print("--- Task 7: Missing Values and Fare Outliers ---\n")

# ==========================================
# 1. Missing-value count, percentage, and plot
# ==========================================
print("1. Missing Values Overview:")
# Calculate counts and percentages
missing_counts = df.isnull().sum()
missing_pct = (missing_counts / len(df)) * 100

# Combine into a DataFrame for a clean display
missing_df = pd.DataFrame({
    'Missing Count': missing_counts,
    'Percentage (%)': missing_pct
})
print(missing_df[missing_df['Missing Count'] > 0]) # Only show columns with missing data
print("\n")

# Plot missing counts as a bar chart
plt.figure(figsize=(8, 5))
missing_counts[missing_counts > 0].plot(kind='bar', color='coral', edgecolor='black')
plt.title('Count of Missing Values per Column')
plt.ylabel('Number of Missing Values')
plt.xlabel('Columns')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# ==========================================
# 2. Fill missing Age values with the mean Age
# ==========================================
print("2. Filling 'Age' with Mean:")
# Count before
missing_age_before = df['Age'].isnull().sum()
print(f"Missing 'Age' values BEFORE filling: {missing_age_before}")

# Fill with mean
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)

# Count after
missing_age_after = df['Age'].isnull().sum()
print(f"Missing 'Age' values AFTER filling:  {missing_age_after}\n")

# ==========================================
# 3. Imputing using Mean, Median, Mode, Random
# ==========================================
# Reloading a fresh copy to demonstrate on original missing data
df_impute = pd.read_csv('train.csv')

print("3. Demonstrating Different Imputation Techniques (on a fresh copy):")

# A. Mean Imputation (Numerical)
df_impute['Age_Mean'] = df_impute['Age'].fillna(df_impute['Age'].mean())
print(f" - Filled Age with Mean: {df_impute['Age_Mean'].isnull().sum()} missing left.")

# B. Median Imputation (Numerical)
df_impute['Age_Median'] = df_impute['Age'].fillna(df_impute['Age'].median())
print(f" - Filled Age with Median: {df_impute['Age_Median'].isnull().sum()} missing left.")

# C. Mode Imputation (Categorical - e.g., Embarked)
# .mode() returns a Series, so we grab the first element [0]
mode_embarked = df_impute['Embarked'].mode()[0]
df_impute['Embarked_Mode'] = df_impute['Embarked'].fillna(mode_embarked)
print(f" - Filled Embarked with Mode ('{mode_embarked}'): {df_impute['Embarked_Mode'].isnull().sum()} missing left.")

# D. Random Value Imputation (Numerical)
# Extract non-null ages to sample from
non_null_ages = df_impute['Age'].dropna()
# Count how many NaNs we need to fill
num_nans = df_impute['Age'].isnull().sum()
# Generate random samples from the existing ages
random_samples = np.random.choice(non_null_ages, size=num_nans, replace=True)

# Create the new column, initially a copy of Age
df_impute['Age_Random'] = df_impute['Age'].copy()
# Assign the random samples specifically to the NaN indices
df_impute.loc[df_impute['Age'].isnull(), 'Age_Random'] = random_samples
print(f" - Filled Age with Random values: {df_impute['Age_Random'].isnull().sum()} missing left.\n")

# ==========================================
# 4. Fare Outliers (IQR Method)
# ==========================================
print("4. Fare Outlier Detection (IQR Method):")

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

# Calculate Interquartile Range (IQR)
IQR = Q3 - Q1

# Calculate Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify Outliers
outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
outliers_count = len(outliers)

print(f" - Q1 (25th percentile): {Q1:.2f}")
print(f" - Q3 (75th percentile): {Q3:.2f}")
print(f" - IQR: {IQR:.2f}")
print(f" - Lower Bound: {lower_bound:.2f}")
print(f" - Upper Bound: {upper_bound:.2f}")
print(f" - Total Outliers Detected: {outliers_count}")
```

**Task 8 - Features and Pivot Table**

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('train.csv')

print("--- Task 8: Features and Pivot Table ---\n")

# ==========================================
# 1. Feature Engineering
# ==========================================
# Create FamilySize: siblings/spouses + parents/children + 1 (themselves)
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Create IsAlone: 1 if FamilySize is 1, else 0
# Using numpy.where is a clean way to apply this conditional logic
df['IsAlone'] = np.where(df['FamilySize'] == 1, 1, 0)

print("1. New Features Added ('FamilySize' and 'IsAlone'):")
# Display a few rows to verify the new columns
print(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head(), "\n")


# ==========================================
# 2. Pivot Table
# ==========================================
# Create pivot table: rows=Sex, columns=Pclass, values=mean Survived
# The default aggfunc in pivot_table is 'mean', but we specify it for clarity
pivot_surv = df.pivot_table(
    index='Sex',
    columns='Pclass',
    values='Survived',
    aggfunc='mean'
)

print("2. Pivot Table (Mean Survival Rate by Sex and Pclass):")
print(pivot_surv, "\n")


# ==========================================
# 3. Find the Highest and Lowest Groups
# ==========================================
# We can use unstack() or stack() to convert the pivot table to a 1D Series
# which makes it easy to find the index of the max and min values.
stacked_pivot = pivot_surv.stack()

highest_group = stacked_pivot.idxmax()
highest_value = stacked_pivot.max()

lowest_group = stacked_pivot.idxmin()
lowest_value = stacked_pivot.min()

print("3. Highest and Lowest Survival Groups:")
print(f" - Highest Survival Rate: {highest_group[0].capitalize()}s in Class {highest_group[1]} ({highest_value * 100:.2f}%)")
print(f" - Lowest Survival Rate:  {lowest_group[0].capitalize()}s in Class {lowest_group[1]} ({lowest_value * 100:.2f}%)")
```
